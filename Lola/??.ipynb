```python
'''
f = t^2 -2k +l
k = 3t - l
l = 2t + 5k -3l

f(0) = 7
k(0) = 3
l(0) = 2

h = 0.25
'''

def f_prim(t, k, l):
    return t**2 - 2*k + l

def k_prim(t, l):
    return 3*t - l

def l_prim(t, k, l):
    return 2*t + 5*k - 3*l


h = 0.25
t = 0

f = [7]
k = [3]
l = [2]

for i in range(2):
    df1 = h*f_prim(t, k[-1], l[-1])
    dk1 = h*k_prim(t, l[-1])
    dl1 = h*l_prim(t, k[-1], l[-1])

    df2 = h*f_prim(t+h, k[-1]+dk1, l[-1]+dl1)
    dk2 = h*k_prim(t+h, l[-1]+dl1)
    dl2 = h*l_prim(t+h, k[-1]+dk1, l[-1]+dl1)

    f.append(f[-1]+(df1+df2)/2)
    k.append(k[-1]+(dk1+dk2)/2)
    l.append(l[-1]+(dl1+dl2)/2)

print(f)
print(k)
print(l)

```
